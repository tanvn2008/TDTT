```python
#Bai_3
a,b,c=map(int,input().split())
if(a<=0 or b<= 0 or c<= 0 or a+b<=c or b+c<=a or a+c<=b):
    print("Khong la ba canh cua mot tam giac")
else:
    if(a==b and b==c and c==a ):
        print("deu")
    elif(a==b or b==c or c==a):
        if(a*a+b*b==c*c or b*b+c*c==a*a or a*a+c*c==b*b):
            print("vuong can")
        else:
            print("can")
    elif(a*a+b*b==c*c or b*b+c*c==a*a or a*a+c*c==b*b):
        print("vuong")
    else:
        print("thuong")
```

```python
#Bai_5
a,b,c=map(int,input().split())
delt = b*b-4*a*c
if(delt<0):
    print("Phuong trinh vo nghiem")
elif(delt>0):
    print(f"Phuong trinh co 2 nghiem la x1={(-b+delt**0.5)/(2*a)} x2={(-b-delt**0.5)/(2*a)}")
else:
    print(f"Phuong trinh co nghiem kep x1=x2={-b/(2*a)}")

```

```python
#Bai_8
a,b,c,d,m,n=map(int,input().split())
delt=a*d-b*c
delx=m*d-b*n
dely=a*n-c*m
if(delt!=0):
    print(f"x={delx/delt} y={dely/delt}")
else:
    if(delx!=0 or dely!=0):
        print("Vo nghiem")
    elif(delx==0 and dely==0):
        print("Vo so nghiem")

```
