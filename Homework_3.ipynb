```python
#Bai_4
n=int(input())
for i in range(1,n+1):
    if(i==1):
        print(" "*(n-2),"*")
    elif(i==n):
        print("*"*(2*n-1))
    else:
        for j in range(1,2*n):
            if(j==n-i+1 or j==n+i-1):
                print("*",end="")
            else:
                print(" ",end="")
        print("")
```

```python
#Bai_14
minn=-1
maxx=-1
while(1):
    x=int(input())
    if(x==-1):
        break
    if(minn==-1):
        minn=x
    else:
        minn=min(minn,x)
    if(maxx==-1):
        maxx=x
    else:
        maxx=max(maxx,x)
print(maxx,minn)
```
